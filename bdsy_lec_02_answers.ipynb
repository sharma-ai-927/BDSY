### **For Loops in Python**

```python
# Motivation 
a_list = ["first", "second", "third", "forth"]
```

Learning `for` loops in Python is a fundamental skill designed to move a programmer from performing manual, repetitive tasks to creating automated, efficient, and concise code. They are essential for iterating over data structures, reducing code redundancy, and handling automation, such as analyzing large datasets or file processing.

```python
# Generic Syntax. (Don't run this cell.)

for {iteration variable} in {iterable}:
           # do things
```

- Each for loop starts with the keyword `for`.

- The `iteration variable` is a variable that is declared only in the scope of the loop that is used to 'point' at the elements in the sequence.

- The `iterable` is a data structure that contains the sequence of elements we are going to perform tasks on, for example a list or a dictionary. 

```python
# 1.A
# Loop over items in a list.
# Write a for loop to print every element of a_list. 

a_list = ["first", "second", "third", "forth"]

for i in a_list:
    print(i)
```

**Remark: When you call range, its value is not a list nor sequence of numbers. Loosely speaking, `range()` saves the instruction to iterate over a range of numbers. For example, `range(5)` defines the instruction to iterate over the first 5 numbers (starting at 0). That is why, used in a loop, it seems as if it is iterating over a list.**

```python
# 1.B
# Compute sum squares of numbers from 1 to 10.
my_sum = 0

for i in range(1,11):
    my_sum = my_sum + (i ** 2)

my_sum
```

# ***Functions***

We have already used many functions in this class that are built into Python or are imported from different modules/packages. 



Writing your own functions in Python is important because it leads to cleaner, more efficient, and easier-to-manage code. Functions help you follow core programming principles, such as modularity, abstraction, and the ***"Don't Repeat Yourself"*** (DRY) principle (Read This: <https://en.wikipedia.org/wiki/Don%27t_repeat_yourself>). 

For example, **Code Reusability:** Functions let you define a task once and call it whenever needed in your program, or even in different projects. This avoids repetitive coding and ensures consistency.

```python
def function_name(parameter1, parameter2=default_value, *args, **kwargs):
    """
    Docstring: This is an optional string that describes what the function does.
    It can span multiple lines.
    """
    # Function body: This is where the code that the function executes resides.
    # It must be indented.
    statement1
    statement2
    # ...
    return expression # Optional: returns a value from the function
```

Explanation of Elements:
- `def` keyword: This keyword is used to declare that a function is being defined.
  
- `function_name`: This is the identifier for the function, following Python's naming conventions (lowercase with underscores for multiple words).

- Parentheses `()`: These enclose the function's parameters (also known as arguments).

  - `parameter1`: A required positional argument.
  - `parameter2=default_value`: An optional argument with a default value. If the caller does not provide a value for this parameter, the default_value will be used.
  - `*args`: Allows the function to accept an arbitrary number of positional arguments, which are collected into a tuple.
  - `**kwargs`: Allows the function to accept an arbitrary number of keyword arguments, which are collected into a dictionary.
    

- Colon `:`: Marks the end of the function header and the beginning of the function body.

- Indentation: The lines of code that form the function's body must be consistently indented (typically with 4 spaces) after the colon. This indentation defines the scope of the function.

- Docstring (optional): A string literal immediately following the function header, enclosed in triple quotes (`"""Docstring"""` or `'''Docstring'''`). It provides documentation for the function and can be accessed via `function_name.__doc__`.

- `return` statement (optional): Used to send a value back to the caller of the function. If return is omitted or used without an expression, the function implicitly returns `None`.

#### 5. Create a function `flip_coins(n, prob)` that generates `n` random numbers (between 0 and 1) and returns the proportion of numbers that are less than `prob`. 

```python
# Wtite your code here. 

def flip_coins(n, prob):
    rand_nums = np.random.rand(n)
    prop = np.mean(rand_nums < prob)
    return prop 



```

#### 6. Write a Python function that takes a string as input and returns a dictionary containing the number of uppercase and lowercase characters in the string. Any characters that cannot be categorized as uppercase or lowercase (e.g., symbols) should be counted as "other".
- Hint: Use may want to use `my_string.isupper()` and `my_string.islower()` methods that we learnt earlier in the semester.
- Hint: The plus-equals operator `+=` provides a convenient way to add a value to an existing variable and assign the new value back to the same variable. 



```python
# Write your code here | Ignore this since we didn't talk about dictionaries. 

def count_upper_lower_chars(input_string):
    char_count = {"uppercase": 0, "lowercase": 0, "other": 0}
    for char in input_string:
        if char.isupper():
            char_count["uppercase"] += 1
        elif char.islower():
            char_count["lowercase"] += 1
        else:
            char_count["other"] += 1
    return char_count
```

```python
# Check your function.

sentence = "Hello, World! I love Yale!"
result = count_upper_lower_chars(sentence)
print(result)   


```

## **Some food for thought: Can we use the `print()` function instead of `return`?**

**Rule of Thumb**

- If the result needs to be used later, use `return`

- If it just needs to be seen, `print` is fine

You cannot use `print()` instead of `return` because they serve fundamentally different purposes in Python: `print()` is for displaying information to a human user, while `return` is for sending a value back to the program's code for further use.


**Code Reusability:** A function that returns a value acts like a self-contained operation that can be a building block for larger, more complex programs. For example, a function that calculates the sum of two numbers should `return` the sum so that the result can be used in subsequent calculations. If it only `print`ed the sum, the program would have no way to access that number internally.

**Error Prevention:** If a function uses `print()` but has no `return` statement, attempting to store its "result" in a variable will assign the value `None` to that variable, which can lead to unexpected errors in your program later on.

```python
# Example 1

def double(x):
    return x * 2

print(double(5) + 3)  # Works
```

```python
# Example 2

def double(x):
    print(x * 2)

print(double(5) + 3)  # Error!
```
